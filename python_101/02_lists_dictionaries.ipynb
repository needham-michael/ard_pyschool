The previous notebook introduced using variables to represent values of different types like strings, integers, floats, and bools. In many cases, we want to keep track of many different values all at once. Python implements different kinds of objects which essentially work as containers that can hold an arbitrary number of values. Two of the key data structures in this category are lists (`list`) and dictionaries (`dict`).

At a basic level,
* __Lists__ are **ordered** collections of values which can be **indexed** to retrieve values
* __Dictionaries__ are **unordered** collections of values which can be retrieved by accessing the value's associated **key**

## Lists

Let's start with a `list`, which is defined using open and closed square brackets `[` and `]`. Items in lists are separated by a comma `,`. Below, we will define a simple list of strings

````python
fruits = ["apple","orange","banana","strawberry"]
fruits
````

### Adding Elements to a List

Once a list is defined, we can use the `<list>.append()` function to add an element to the end of the list

````python
fruits = ["apple","orange","banana","strawberry"]

fruits.append("watermelon")
fruits.append("kiwi")

fruits
````

### Accessing by Index

Above, we said that lists are **ordered** which means that once an item is added to a list it is always in that same position, unless we explicitly change it. This is in contrast to dictionaries, as we will see later on.

Since the list is **ordered** we can retrieve an item from within a list as long as we know which place it is in (e.g., first, second, third, and so on). This is called **indexing** the list. To do so we again use open and closed square brackets `[` and `]` along with an integer which describes the placement of the value we care about from within the list.

````python
fruits = ["apple","orange","banana","strawberry","watermelon","kiwi"]
fruits[1]
````

Do you notice anything odd about the cell above? Here, we have indexed our list `fruits` at position `1`, which gave us the value `"orange"`, even though 

The reason for this is that python uses __[zero-based indexing](https://en.wikipedia.org/wiki/Zero-based_numbering)__, which means that the initial element of a sequence is assigned the index `0` instead of `1`. Don't worry too much about why this is the case, it's simply a convention of using python.

So to get the initial element of the list, we index the list at position `0`:

````python
fruits = ["apple","orange","banana","strawberry","watermelon","kiwi"]
fruits[0]
````

We can use **slicing** to extract consecutive elements from a list. The syntax is to specify the `start` and `stop`, which are the indices where the slice should begin and end, respectively. These indices are placed within square brackets separated with a colon `:`. 

Note that the `start` index is inclusive, while the `stop` index is exclusive.  Essentially, we retrieve items if the index is greater than or equal to `start`, and strictly less than `stop`

> This is sometimes described as a "Half-open interval `[start, stop)` where the open square bracket `[` indicates that `start` index is included, while the closed parenthesis `)` indicates that the `stop` index is excluded" 

````python
# retrieve elements at the indices bounded by 1 (inclusive) and 4 (exclusive)
start = 1
stop = 4

# Index         0       1         2         3            4          5
# Slice(1,4)            [--------------------------------)
# Included?     N       Y         Y         Y            N          N
fruits =    ["apple","orange","banana","strawberry","watermelon","kiwi"]
fruits[start:stop] 
````

If either `start` or `stop` are excluded, the slice will return all elements starting at the beginning, or extending to the end of the list, respectively.

````python
fruits = ["apple","orange","banana","strawberry","watermelon","kiwi"]

# Retrieve the first three elements of the list
fruits[:3] 
````

If we try to retrieve a value from our list at an index that does not exist (for example, if our list has `6` elements and we try to index at position `73`), we will get an `IndexError`:

````python
fruits[73]
````

### Operations on Lists

Python implements certain built-in functions that can perform operations on lists. There are many of these, but important ones to know are:
- `min`: return the item with the smallest value
- `max`: return the item with the largest value
- `len`: return the length of the list
- `sum`: return the sum of the elements of the list

````python
my_list = [8,6,7,5,3,0,9]

print("Min: ",min(my_list))
print("Max: ",max(my_list))
print("Sum: ",sum(my_list))
print("Length: ",len(my_list))

# Combine sum and len to calculate the average (mean)
print("Average: ",sum(my_list)/len(my_list))
````

## Dictionaries

Dictionaries are similar to lists in that they allow us to keep track of multiple values. But instead of assigning each value to an ordered position, we instead associate each value with some keyword. This is often refered to as a __`key-value pair`__

We define a list using open and closed curly braces `{` and `}` where a colon `:` separates a key from its value, and commas `,` separate different `key-value pairs`: 

```python
my_dict = {"key1":val1, "key2":val2}
```


````python
# For readability, can separate key-value pairs onto different lines
student_grades = {
    "Alice":91,
    "Bob":83,
    "Charlie":79,
    "David":89,
    "Erin":95,
    "Frank":85,
    "Grace":88,
    "Heidi":82,
}
````

### Accessing by Key

As mentioned above, we have associated each grade (i.e., the `value` of the `key-value pair`) with a student's name (i.e., the `key`). To retrieve the grade, we will once again use open and closed square brackets `[` and `]` along with the specific `key` of the `key-value pair`.

````python
student = "Grace"
grade = student_grades[student]

print("Student:",student)
print("Grade:",grade)
````

Similar to the `IndexError` we received above, if we try to use a key that is not in our dictionary, we will get a `KeyError`

````python
student = "Ivan"
grade = student_grades[student]

print("Student:",student)
print("Grade:",grade)
````

## Combining Lists and Dictionaries

Both lists and dictionaries are **nestable** which means that that elements of a list or dictionary can be lists or dictionaries themseleves. Additionally, you can combine lists with dictionaries, as in the example below where each student in our dictionary has a list of their grades for six different assignments:

````python
student_grades = {
    "Alice":[99, 77, 92, 71, 100, 88],
    "Bob":[97, 96, 99, 89, 96, 76],
    "Charlie":[79, 89, 66, 84, 84, 82],
    "David":[85, 85, 86, 66, 80, 77],
    "Erin":[73, 70, 92, 94, 70, 81],
    "Frank":[75, 99, 86, 79, 69, 75],
    "Grace":[94, 95, 72, 85, 93, 75],
    "Heidi":[94, 66, 90, 79, 68, 98],
}
````

We can combine key access to our dictionary with operations on the list of grades to report the performance of each student.

````python
student = "Bob"
grades = student_grades[student]

print("Student:",student)
print("Grades:",grades)
print("   #Grades :",len(grades))
print("   Max     :",max(grades))
print("   Min     :",min(grades))
print("   Average :",sum(grades) / len(grades))
````
