# Variables, Types, and Operators

A computer program can be thought of as a set of instructions which create, manipulate, and retrieve `variables`. A variable is simply a descriptive name used to keep track of some particular value. These values can be of different `types` such as numeric types (integers and decimal numbers such as `5` and `3.14159`), string types

In python, the equal sign `=` is the `assignment operator` which is used to store some value as a `variable` for later use.

### Variables

In the example below, we will assign the value of 5 to a variable named `var1`, and then use the `print` function to display the value.





````python
var1 = 5

# Use `print` to show the value of the variable
print(var1)
````

> __A quick note on `print`__
>
> In the cell above, the second code line (under the comment line) uses the `print` function to display the value of `var1`. As in many programming languages, python implements a `print` function (a holdover from the days when a computer needed to literally print its output onto paper) to take some text and write it to the screen for the programmer to read. `print` is one of the fundamental ways that a programmer interacts with their program and is also the simplest way to begin to __[debug](https://en.wikipedia.org/wiki/Debugging)__ code. 
>
> In the cells below, we will continue to "call" the `print` function to inspect the values of our variables. The __[syntax](https://en.wikipedia.org/wiki/Python_syntax_and_semantics)__ to use the `print` function is to write the `print` keyword, followed by an open parenthesis `(`, followed by our desired printout content, and then end the function call with a closed parenthesis `)`. All together, it looks something like this: `print(`*`content`*`)`
 
Ok, back on track. Once the value of a variable is set, it does not change... unless we tell it to

````python
var1 = 5
print(var1)
print(var1)
print(var1)

# Now, update the value...
var1 = 10
print(var1)
````

Arithmetic operations (and all other operations, too) work *identically* between variables and their underlying values

````python
var1 = 10
var2 = 3

# Using numbers
print(10 + 3)

# Using variables
print(var1 + var2)

# And a mix of both
print(var1 + 3)
````

### Types

Python implements a variety of different data `types` which store and manage different kinds of data. Some basic types to understand are:

* strings (`str`): literal representations of characters, surrounded by single or double quotation marks (i.e., `'` or `"`) 
* booleans (`bool`): logical values which can only be `True` or `False`
* integers (`int`): whole numbers (positive, negative, or zero)
* floating point numbers (`float`): decimal point numbers

> __On `floats`__
>
> For our purposes here, thinking of floats as decimal representations is fine, but there is a whole interesting thread of computer science about this, which you can read about in this wikipedia article about  __[floating-point arithmetic](https://en.wikipedia.org/wiki/Floating-point_arithmetic)__. In short, if a computer can only represent a number based on a fixed number of ones and zeros, how do you design a system that allows a user to represent numbers across a wide range of scales from the very large to the very small?

If you are ever in doubt, you can use the built-in `type` function to find out the type of some variable

````python
print(type("123"))
print(type(True))
print(type(123))
print(type(1.23))
````

For the numeric types, python will automatically convert integers to floats if necessary to allow a program to proceed (some other programming languages will give a user an error if they try to add, e.g., `1 + 2.0`)

````python
# Integer + Integer = Integer
print(5+3)

# Integer + Float = Float
print(5+3.0)

# Float + Float = Float
print(5.0+3.0)
````

Sometimes, operations between variables will give an error if the types are not correct...

````python
# This evaluates fine
print(5*5)

# This throws an error
print("5"*"5")
````

__Be Careful! Accidentally mixing data types can lead to unintended results!__

In the case below, the program doesn't throw an error and instead gives you a result... just not the one you were expecting!

````python
# Multiplying integers gives an integer
print(5*5)

# But multiplying an integer by a string gives you copies of the string!
print(5*"5")
````

Boolean (`bool`) types allow your program to make decisions about what lines of code should or should not be executed based on a condition. The next notebook will discuss this and other aspects of __[control flow](https://en.wikipedia.org/wiki/Control_flow)__ in python, but the usefulness of `bool` types can be illustrated with the concept of a simple `if / else` block:



````python
# Define a variable called `condition` with a `bool` type
condition = False

# Only one of the indented code blocks will execute, based on whether the condition
# variable is `True` or `False`

if condition:
    print(condition)
    print("The condition is met!")
else:
    print(condition)
    print("The condition is not met!")
````

### Operators

The `if / else` block is a nice segue into the concept of `operators` in python. Operators are reserved symbols or short keywords used to perform operations on and between variables. We have already used the assignment operator `=` to assign a value to a variable.

Comparison operators compare two values an return either `True` or `False`:

* Equal to: __`==`__
* Not equal to: __`!=`__
* Greater than: __`>`__
* Less than: __`<`__
* Greater than or equal to: __`>=`__
* Less than or equal to: __`<=`__


````python
print(5 == 5) # 5 is `equal` to 5 -> True
print(4 != 5) # 4 is `not equal` to 5 -> False
print(23 != 23) # 23 is not `not equal` to 23 -> False (Read this one again if you need to!)
````

Arithmetic operators perform mathematical calculations between two values

* Addition: __`+`__
* Subtraction: __`-`__
* Multiplication: __`*`__
* Division: __`/`__
* Floor Division$^*$: __`//`__
* Modulus$^*$$^*$: __`%`__

*$^*$ Floor division divides two numbers and rounds the result down to the nearest integer*  
*$^*$$^*$if you have not come across modular arithmetic before, ignore this for now... although you use this every day when telling time :)*

````python
print(5+5)
print(10-8)
print(12*12)

# Regular division gives a float...
print(714/25)

# Floor division rounds town to the nearest integer...
print(714//25)
````

````python
# Lastly, modular arithmetic gives the remainder after division
print(714%25)
````

There are a number of other operators used in python, but keeping with the comparison and arithmetic operators is fine for just starting.
