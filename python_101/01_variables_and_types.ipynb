# Variables, Types, and Operators

A computer program can be thought of as a set of instructions which create, manipulate, and retrieve `variables`. A variable is simply a descriptive name used to keep track of some particular value. These values can be of different `types` such as numeric types (integers and decimal numbers such as `5` and `3.14159`), string types

In python, the equal sign `=` is the `assignment operator` which is used to store some value as a `variable` for later use.

### Variables

In the example below, we will assign the value of 5 to a variable named `var1`, and then use the `print` function to display the value.





```python
var1 = 5

# printout the value of the variable
print(f"The value of var is {var1}")
```

Once the value of a variable is set, it does not change until we tell it to

```python
var1 = 5
print(f"1. var1 = {var1}")
print(f"2. var1 = {var1}")
print(f"3. var1 = {var1}")

var1 = 10
print(f"4. var1 = {var1}")
```

Arithmetic and other operations works *identically* between variables and their underlying values

```python
var1 = 10
var2 = 3

print(f"Using numbers: 10 + 3 = {10 + 3}")
print(f"Using variables: var1 + var2 = {var1 + var2}")
print(f"And a mix of both: var1 + 3 = {var1 + 3}")
```

### Types

Python implements a variety of different data `types` which store and manage different kinds of data. Some important types to consider are:


* strings (`str`): literal representations of characters, surrounded by single or double quotation marks (i.e., `'` or `"`) 
* integers (`int`): whole numbers (positive, negative, or zero)
* floating point numbers (`float`): decimal point numbers
* booleans (`bool`): logical values which can only be `True` or `False`
