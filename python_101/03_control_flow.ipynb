## Decision making with `If / Else` blocks
In the previous notebook, we introduced the concept of an `if / else` block:

```python
if <EXPRESSION>:
    # statement if EXPRESSION is True
else:
    # statement if EXPRESSION is False
```

You can think of an `if / else` block as structuring a computer program like a decision tree or a flow chart. As the code progresses, we use these statements with the `if` and `else` keywords to decide which lines of code are (not) executed under different conditions.

#### Positive, Negative, or Zero

To illustrate this concept in a simple way, let's write a short program to characterize numbers as Positive, Negative, or Zero...

...however since there are three categories we will need to use **two** `if / else` blocks

````python
# Select a number
number = -153

# First, check if the number is greater than zero
if number > 0:
    print(number,"is positive.")

else:
    # Then, check if the number is less than zero
    if number < 0:
        print(number,"is negative.")

    # Otherwise, the number must be equal to zero
    else:
        print(number,"is zero.")
    
````

Above, we had to use a second **nested** `if / else` block to differentiate between negative numbers and zero. This type of situation where a program needs to check multiple conditions happens frequently enough that python has an additional keyword `elif` that we can use to shorten this code.

An `if / elif / else` block has the following form:

```python
if <EXPRESSION_A>:
    # statement if EXPRESSION_A is True
elif <EXPRESSION_B>:
    # statement if EXPRESSION_A is False but EXPRESSION_B is True
else:
    # statement if EXPRESSION_A and EXPRESSION_B are both False
```

````python
# Select a number
number = -153

# First, check if the number is greater than zero
if number > 0:
    print(number,"is positive.")

# Then, check if the number is less than zero
elif number < 0:
        print(number,"is negative.")

# Otherwise, the number must be equal to zero
else:
    print(number,"is zero.")
````

You can have an arbitrary number of `elif` statements underneath the first `if` statement, and the first one of these that evaluates to `True` will run

````python
number = 3

if number < 0:
    print("Number is less than 0")
elif number < 1:
    print("Number is less than 1")
elif number < 2:
    print("Number is less than 2")
elif number < 3:
    print("Number is less than 3")
elif number < 4: # <--- If number = 3, then `number < 4` is True
    print("Number is less than 4")
elif number < 5:
    print("Number is less than 5")
````

## Defining `functions`

The code we wrote previously does a good what we want, but it might be useful to save this code and use it later. We can do this by defining a `function` using the `def` keyword with the following syntax

```python
def function_name(): # <--- note the final colon ":" at the end of the line
    # code to run

# execute the function
function_name()
```

A function may (or may not) have arguments, and it may (or may not) return some values back to the main program using the `return` keyword

```python
def add_numbers(num1,num2):
    value = num1 + num2
    return value

num1 = 5
num2 = 3
value = add_numbers(num1,num2)
print(num1,"+",num2,"=",value) # evaluates to "5 + 3 = 8"
```

Let's wrap up the code we wrote above in a function that we can use more easily

````python
def pos_neg_or_zero(number):
    # First, check if the number is greater than zero
    if number > 0:
        print(number,"is positive.")
    
    # Then, check if the number is less than zero
    elif number < 0:
            print(number,"is negative.")
    
    # Otherwise, the number must be equal to zero
    else:
        print(number,"is zero.")


# Now, we can call this function multiple times
pos_neg_or_zero(-100)
pos_neg_or_zero(-50)
pos_neg_or_zero(-10)
pos_neg_or_zero(-5)
pos_neg_or_zero(0)
pos_neg_or_zero(5)
pos_neg_or_zero(10)
pos_neg_or_zero(50)
pos_neg_or_zero(100)
````

## Repeating code with a `for` loop

The previous example illustrates that sometimes we will repeat the same code multiple times in the same program. We can use a programming concept called a  __[loop](https://en.wikipedia.org/wiki/Loop_(statement))__ repeatedly call the same code as many times as we want.

Python implements **two** different types of loops. Here, we will use the first of these which is a `for` loop to call the above function multiple times.

> The second type is a `while` loop, that continues to evaluate the same code until a condition is met, at which point the program exits the loop. 

The syntax of a `for` loop is:
```python
for value in <iterable>: # <--- again, note the final colon ":"
    # code to be repeated each iteration
```

Here, the `<iterable>` can be a `list`, a `dict`, or some other collection of values.

````python
numbers_to_check = [-100,-50,-10,-5,0,5,10,50,100]
for number in numbers_to_check:
    pos_neg_or_zero(number)
````
