## Puzzle #1 Solution

````python
def fizzbuzz(number):
    output = ""
    if number % 3 == 0:
        output += "Fizz"
    if number %5 == 0:
        output += "Buzz"
    if output == "":
        return number
    return output

for number in range(1,101):
    output = fizzbuzz(number)
    print(output)
````

### Puzzle #2 Code Solution

````python
for threshold in [5, 20,500,100000]:
    
    counter = 0
    number = 1
    fizzbuzz_numbers = []
    
    while counter < threshold:
    
        output = fizzbuzz(number)
    
        if type(output) == str:
            fizzbuzz_numbers.append(number)
            counter += 1
    
        number += 1

    print("The SUM of the first",threshold, "fizzbuzz numbers is:", sum(fizzbuzz_numbers))
    
    if len(fizzbuzz_numbers) <= 10:
        print(fizzbuzz_numbers)
    
    print("-"*79)
````
