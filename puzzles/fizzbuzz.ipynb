# Fizz buzz

__[Fizz buzz](https://en.wikipedia.org/wiki/Fizz_buzz)__ is a counting word game with simpile rules:

1. Count numbers incrementally starting with one up to some upper limit
2. If a number is divisible by `3`, replace it with `Fizz`
3. If a number is divisible by `5`, replace it with `Buzz`
4. If a number is divisilbe by both `3` and `5`, replace it with `Fizz Buzz`

## Puzzle #1
Write a python program to output to print the Fizz Buzz numbers starting at `1` up to and including `50`

## Puzzle #2
What is the sum of the first 5 FizzBuzz numbers? The first 500? The first 50,000?

## Puzzle #1 Solution

```python
def fizzbuzz(number):
    output = ""
    if number % 3 == 0:
        output += "Fizz"
    if number %5 == 0:
        output += "Buzz"
    if output == "":
        return number
    return output

for number in range(1,51):
    output = fizzbuzz(number)
    print(output)
```

## Puzzle #2 Solution

| N | Sum |
| - | --- |
| 5 | 33 |
| 500 | 268,394 | 
| 50,000 | 2,678,625,000 |

### Puzzle #2 Code Solution

```python
threshold = 500

for threshold in [5,500,50000]:
    
    counter = 0
    number = 1
    fizzbuzz_numbers = []
    
    while counter < threshold:
    
        output = fizzbuzz(number)
    
        if type(output) == str:
            fizzbuzz_numbers.append(number)
            counter += 1
    
        number += 1

    print("The SUM of the first",threshold, "fizzbuzz numbers is:", sum(fizzbuzz_numbers))
    
    if len(fizzbuzz_numbers) <= 10:
        print(fizzbuzz_numbers)
    
    print("-"*79)
```
