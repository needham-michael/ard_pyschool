## 1. Read a file of (phony) air quality data

````python
def read_data(filename):
    """Custom function to read the phony air quality data files used in this exercise

    Parameters
    ----------
    filename : str
        Path to an air quality data file 

    Returns
    -------
    timestep, conc : list
        Lists containing the floating point values representing the timestep and 
        air quality concentration used in this exercise
    """
    
    # Initialize the timestep and concentration as empty arrays
    timestep = []
    conc = []

    with open(filename,mode='r') as file:
        for line in file.readlines():
            line = line[:-1] # remove the trailing newline "\n" character

            values = line.split(",")

            timestep.append(float(values[0]))
            conc.append(float(values[1]))

    return timestep, conc
   
````

````python
timestep_01, conc_01 = read_data("./ExampleO3_Monitor01.txt")

# Printout the first 24 elements of each array
for ix in range(0,24):
    print(f"{ix:>5} | {timestep_01[ix]:>5} | {conc_01[ix]}")
print("...")
````

## 2. Convert the air concentrations from ppm to ppb

From the printout above, it is clear that our data are given in parts per million (ppm), while O3 data is often presented in parts per billion (ppb). It is easy to convert between the two, by simply multiplying the data in ppm by `1000`:

$$
\text{CONC}_{ppb} = 1000 \times \text{CONC}_{ppm}
$$

To do this in python we need to:
1. __*Iterate*__ over each element in our array using a `for` loop
2. Multiply the value of each element by our conversion factor (i.e., `1000`)
3. Store the output in a new array, representing data in parts per billion

Here is some __[pseudocode](https://en.wikipedia.org/wiki/Pseudocode)__ of how this could work:

```python
new_array = []
for value in array:
    new_value = value * conversion_factor
    new_array.append(new_value)
```
