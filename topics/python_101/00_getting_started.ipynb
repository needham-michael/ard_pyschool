```python
# This is a (python) code cell. 

# In python the "#" character starts a "comment line" which is ignored by the 
# interpreter when the computer executes the code. Comments are extremely useful to
# help future users to understand what a particular line in a program is doing.

# Also, 9 times out of 10, that future user is YOU, so write good comments! 


# ======================================================================================
# --------------------------------- Fibonacci Sequence ---------------------------------
# ======================================================================================
# This code snippet calculates and writes out the first N numbers of the famous 
# fibonacci sequence. Feel free to change the "number of numbers" that the program will
# show by updating the `numbers_to_show` variable to some number other than the default
# of 20.
# --------------------------------------------------------------------------------------

numbers_to_show = 20


counter = 1
x0 = 0
x1 = 1

print("="*20)
print("Fibonacci Numbers")
print("="*20)

# For loop to generate the print the first N numbers of the fibonacci sequence
for counter in range(numbers_to_show):
    if counter == 0:
        fib = x0
    elif counter == 1:
        fib = x1
    else:
        fib = x0 + x1

        x0 = x1
        x1 = fib
        
    print(f"{counter:<2} | {fib}")
```
